## Left shift operations

```python
x=15
x<<=1
print(x)

#before binary 1111
# after binary 11110
#left shift (<<): move bits left and add bit position on right 0
```

```python
x=19  #10011
x<<=2 #1001100
print(x)
```

## Right left Operation

```python
x=15
x>>=1
print(x)

#>>1
#before binary 1111
#after binary 01111
```
