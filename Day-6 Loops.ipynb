```python
# for loop we can use iterate over sequence of items,list,tuple,range,dictionary etc
a="python"
for i in a:
    print(i)
```

```python
number=["aws","azure","gcp"]
for i in number:
    print(i)
```

```python
number=[1,2,3,4,5,6]
for i in number:
  if i==3:
    print(i)
```

```python
number=[1,2,3,4,5,6]
for i in number:
  if i==3:
      break
  print(i)
```

```python
# for loop with dictionary
number={"name":"veera",
       "course":"fde",
       "fee":14000,
       "location":"online"}
for i in number.items():
    print(i)
```

```python
for i in number.values():
    print(i)
```

```python
for i in number.keys():
    print(i)
```

```python
total=1
dev=range(1,5)
for i in dev:
    total=total+i
print("Total",total)
```
